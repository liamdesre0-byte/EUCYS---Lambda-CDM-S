# ΛCDM+S example analysis

This notebook calls `lcdm_plus_s.solve_background` — the same integrator used by the CLI and Streamlit app. It does not load unpublished MCMC chains.

```python
from lcdm_plus_s import BackgroundParams, solve_background, logistic_weight
from lcdm_plus_s.bayesian_validation import lcdm_recovery_test, KMSMPC_TO_INVGYR

params = BackgroundParams()  # Table 1 fiducials in code
sol = solve_background(params, nsteps=2000)
lcdm = solve_background(BackgroundParams(lcdm_limit=True), nsteps=2000)

print(f"t0 LCDM+S = {sol.t0:.4f} Gyr, chi0 = {sol.chi0:.4f}")
print(f"t0 LCDM   = {lcdm.t0:.4f} Gyr")
print(f"chi(t_crit) = {logistic_weight(params.t_crit_gyr, params.k_gyr, params.t_crit_gyr):.3f}")
print(f"H0 check: {sol.hubble_of_z(0.0):.6f} vs {params.H0_kms_mpc * KMSMPC_TO_INVGYR:.6f} Gyr^{-1}")
print(lcdm_recovery_test(nsteps=1500)["passed"])
```
